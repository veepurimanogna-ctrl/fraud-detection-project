```python
import pandas as pd
```

```python
train_df = pd.read_csv("data/fraudTrain.csv")
test_df = pd.read_csv("data/fraudTest.csv")

print(train_df.shape)
print(test_df.shape)
```

```python
train_df['trans_date_trans_time'] = pd.to_datetime(
    train_df['trans_date_trans_time']
)

test_df['trans_date_trans_time'] = pd.to_datetime(
    test_df['trans_date_trans_time']
)
```

```python
train_df['transaction_hour'] = (
    train_df['trans_date_trans_time'].dt.hour
)

train_df['transaction_day'] = (
    train_df['trans_date_trans_time'].dt.day
)

train_df['transaction_month'] = (
    train_df['trans_date_trans_time'].dt.month
)

train_df['transaction_dayofweek_num'] = (
    train_df['trans_date_trans_time'].dt.dayofweek
)
```

```python
test_df['transaction_hour'] = (
    test_df['trans_date_trans_time'].dt.hour
)

test_df['transaction_day'] = (
    test_df['trans_date_trans_time'].dt.day
)

test_df['transaction_month'] = (
    test_df['trans_date_trans_time'].dt.month
)

test_df['transaction_dayofweek_num'] = (
    test_df['trans_date_trans_time'].dt.dayofweek
)
```

```python
train_df['is_weekend'] = (
    train_df['transaction_dayofweek_num'] >= 5
).astype(int)
```

```python
import numpy as np
```

```python
train_df['amt_log'] = np.log1p(train_df['amt'])
```

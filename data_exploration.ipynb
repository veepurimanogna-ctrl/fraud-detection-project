```python
train_df = train_df.drop_duplicates()
```

```python
train_df['trans_date_trans_time'] = pd.to_datetime(
    train_df['trans_date_trans_time']
)
```

```python
train_df['transaction_hour'] = (
    train_df['trans_date_trans_time'].dt.hour
)
```

```python
train_df['transaction_day'] = (
    train_df['trans_date_trans_time'].dt.day
)
```

```python
train_df['transaction_month'] = (
    train_df['trans_date_trans_time'].dt.month
)
```

```python
train_df['is_fraud'].value_counts()
```

```python
train_df['is_fraud'].value_counts(normalize=True) * 100
```

```python
sns.countplot(x='is_fraud', data=train_df)

plt.title('Fraud vs Non-Fraud Transactions')
plt.xlabel('Fraud (0 = No, 1 = Yes)')
plt.ylabel('Number of Transactions')
plt.show()
```

```python
train_df['amt'].describe()
```

```python
sns.histplot(train_df['amt'], bins=50)

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount')
plt.show()
```

```python
category_fraud = train_df.groupby('category')['is_fraud'].mean() * 100

category_fraud.sort_values(ascending=False)
```

```python
category_fraud.sort_values(ascending=False).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Fraud Rate by Merchant Category')
plt.xlabel('Merchant Category')
plt.ylabel('Fraud Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

```python
hourly_fraud = train_df.groupby(
    'transaction_hour'
)['is_fraud'].mean() * 100

hourly_fraud
```

```python
hourly_fraud.plot(
    kind='line',
    figsize=(10,5),
    marker='o'
)

plt.title('Fraud Rate by Transaction Hour')
plt.xlabel('Hour')
plt.ylabel('Fraud Rate (%)')
plt.grid()
plt.show()
```

```python
gender_fraud = train_df.groupby('gender')['is_fraud'].mean() * 100

gender_fraud
```

```python
gender_fraud.plot(kind='bar')

plt.title('Fraud Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Fraud Rate (%)')
plt.show()
```
